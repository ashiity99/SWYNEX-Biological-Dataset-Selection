# SWYNEX Task 3 — Visualization and Interpretation

## Dataset
Escherichia coli K-12 MG1655 complete reference genome

NCBI Accession: NC_000913.3

This notebook visualizes the nucleotide composition and GC/AT content calculated during Task 2 and interprets the results scientifically.

```python
# Task 2 results used for Task 3 visualization

nucleotides = ['A', 'T', 'G', 'C']
counts = [1142742, 1141382, 1177437, 1180091]

genome_length = 4641652
gc_content = 50.79
at_content = 49.21
ambiguous_bases = 0

print(f"Genome length: {genome_length:,} bp")
print(f"GC content: {gc_content:.2f}%")
print(f"AT content: {at_content:.2f}%")
print(f"Ambiguous bases: {ambiguous_bases}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.bar(nucleotides, counts)

plt.title("E. coli K-12 MG1655 Nucleotide Composition")
plt.xlabel("Nucleotide")
plt.ylabel("Number of Bases")

plt.tight_layout()
plt.show()
```

### Interpretation

G and C are slightly more abundant than A and T. This is consistent with the calculated GC content of 50.79% and AT content of 49.21%.

The four nucleotide types are all present in substantial amounts, and the cleaned sequence contains no ambiguous N bases.

```python
composition_labels = ['GC', 'AT']
composition_values = [gc_content, at_content]

plt.figure(figsize=(7,5))

plt.bar(composition_labels, composition_values)

plt.title("GC vs AT Content")
plt.xlabel("Base Composition")
plt.ylabel("Percentage (%)")
plt.ylim(0, 60)

plt.tight_layout()
plt.show()
```

### Interpretation

The genome has 50.79% GC content and 49.21% AT content, showing a nearly balanced overall base composition with a small GC excess.

# What the Results Do and Do Not Prove

### What the analysis shows

- The reference genome contains 4,641,652 base pairs.
- GC bases make up 50.79% of the sequence.
- AT bases make up 49.21% of the sequence.
- G and C occur slightly more often than A and T.
- The cleaned sequence contains 0 ambiguous N bases.

### What the analysis does NOT prove

- Base composition alone does not prove the function of genes or proteins.
- GC content alone does not establish evolutionary relationships.
- These overall statistics do not show where genes, regulatory regions, or other genomic features are located.
- Nucleotide composition alone does not determine whether a genome is biologically "healthy" or more functional.

### Conclusion

The visualizations provide a clear summary of the nucleotide composition of the E. coli K-12 MG1655 reference genome. The sequence is approximately balanced between GC and AT bases, has a GC content of 50.79%, and contains no ambiguous bases in the cleaned sequence.

These results describe sequence composition and quality, but further genomic analysis would be required to infer gene function, biological mechanisms, or evolutionary history.
